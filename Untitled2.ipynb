# Что такое **DataFrame**

```python
import pandas as pd
import numpy as np

np.random.seed(45)
data={
    "Возраст":np.random.randint(low=18,high=90,size=100),
    "Диагноз":["Гипертония", "Простуда", "Артрит", "Сердечная недостаточность"]*25,
    "Кол. процедур":np.random.randint(low=1,high=10,size=100),
}
df=pd.DataFrame(data)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x="Диагноз", data=df)
plt.title("частота встречаемости болезней")
plt.show
```

```python
filter_df = df[(df["Возраст"] > 60) & (df["Диагноз"] == "Гипертония")]
print(filter_df)
```

```python
df["Возраст"]=df["Возраст"].astype(float)
df.dtypes
```

```python
df["Процедуры на возраст"]=df["Кол. процедур"]/df["Возраст"]
df.head()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.to_csv("data.csv")
```
