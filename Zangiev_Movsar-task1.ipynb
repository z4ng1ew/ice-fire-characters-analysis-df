# 1. Создание датафрейма с данными о персонажах "Песни Льда и Огня"

```python
import pandas as pd
```

```python
# В данных использованы состояния персонажей на конец книги "Танец с драконами" (пятая книга цикла)
# Шестая книга «Песнь льда и огня» Джорджа Р. Р. Мартина под названием «Ветра зимы» ещё не опубликована.
data = {
    "Имя": [
        "Джон Сноу", "Дейнерис Таргариен", "Тирион Ланнистер", "Санса Старк",
        "Арья Старк", "Церсей Ланнистер", "Бран Старк", "Джейме Ланнистер",
        "Теон Грейджой", "Мирцелла Баратеон"
    ],
        "Возраст": [18, 17, 35, 15, 12, 38, 10, 37, 20, 13],
    "Дом": [
        "Старк", "Таргариен", "Ланнистер", "Старк",
        "Старк", "Ланнистер", "Старк", "Ланнистер",
        "Грейджой", "Баратеон"
    ],
    "Жив": [
        True, True, True, True,
        True, False, True, False,
        False, False
    ]
}
df = pd.DataFrame(data)
```

# 2. Вывод информации

```python
print(df)
```
